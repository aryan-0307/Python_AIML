```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
```

```python
df = pd.read_csv('train.csv',usecols=['Age','Fare','Survived'])
df.head()
```

```python
X=df.iloc[:,1:3]
y=df['Survived']
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
X_train.head()
```

```python
si = SimpleImputer()
X_train_missing = si.fit_transform(X_train)
X_test_missing = si.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
lr.fit(X_train_missing,y_train)
y_pred=lr.predict(X_test_missing)
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

```python
from sklearn.impute import MissingIndicator
mi = MissingIndicator()
X_train_missinggg = mi.fit_transform(X_train)
X_test_missinggg = mi.transform(X_test)
X_train['Age_Na']=X_train_missinggg
X_train
```
