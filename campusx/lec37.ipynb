```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('trains.csv',usecols=['GarageQual','FireplaceQu','SalePrice'])
df.head()
```

```python
df.isnull().mean()
```

```python
df['GarageQual'].value_counts().plot(kind='bar')
```

```python
df.mode()
```

```python
fig, ax =plt.subplots(1,1)

df[df['GarageQual'] == 'TA']['SalePrice'].plot(kind='kde', ax=ax, label='TA')
df[df['GarageQual'].isnull()]['SalePrice'].plot(kind='kde', ax=ax, color='red', label='Missing')

ax.legend()
plt.show()
```

```python
temp=df[df['GarageQual'] == 'TA']['SalePrice']
```

```python
temp.head(5)
```

```python
df['GarageQual'].fillna('TA',inplace=True)
```

```python
df['GarageQual'].value_counts().plot(kind='bar')
```

```python
fig ,ax =plt.subplots(1,1)

temp.plot(kind='kde',label='duplicate')

df[df['GarageQual'] == 'TA']['SalePrice'].plot(kind='kde',ax=ax,label='after')

plt.legend()
plt.show()
```
