```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('titanic.csv')
df.head()
```

## OPERATION ON NUMBER COLUMN

```python
df['number'].value_counts().plot(kind='bar')
plt.show()
```

```python
df['Number_Numerical']=pd.to_numeric(df['number'],errors='coerce',downcast='integer')
```

```python
df["Number_Categorical"]=np.where(df['Number_Numerical'].isnull(),df['number'],np.nan)
```

```python
df.head()
```

```python
df.head()
```

## OPERATION ON CABIN COLUMN

```python
df['Cabin_Numerical']=df['Cabin'].str.extract('(\d+)')
df['Cabin_Categorical']=df['Cabin'].str[0]
```

```python
df.head()
```

```python
df.drop(columns='Cabin')
```

## OPERATION ON TICKET

```python
# ✅ Extract numerical part
df['Ticket_numerical'] = df['Ticket'].str.extract(r'(\d+)')

# ✅ Extract categorical part (full text, not just first char)
df['Ticket_categorical'] = df['Ticket'].str.extract(r'([A-Za-z/\.]+)')
```
