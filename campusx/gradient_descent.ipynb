```python
import numpy as np
import pandas as pd
```

```python
from sklearn.datasets import make_regression
```

```python
X,y = make_regression(n_samples=4, n_features=1, n_informative=1, n_targets=1,noise=80,random_state=13)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(X,y)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
reg = LinearRegression()
```

```python
reg.fit(X,y)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='red')
```

## apply gradient descent

```python
y_pred = ((78.35 * X)+100).reshape(4)
```

```python
plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='red',label='OLS')
plt.plot(X,y_pred,color="green",label='b=0')
plt.legend()
plt.show()
```

```python
m=78.35
b=100
```

```python
loss_slope = -2 * np.sum((y- m*X.ravel()-b))
```

```python
loss_slope
```

```python
lr=0.1
```

```python
b_new = b - lr*loss_slope
```

```python
b_new
```

```python
y_pred1 = ((m *X) + b_new).reshape(4)
```

```python
plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='red',label='OLS')
plt.plot(X,y_pred1,color='blue',label="new")
plt.plot(X,y_pred,color="green",label='b=0')
plt.legend()
plt.show()
```

```python
m=78.35
b=-100
lr=0.01

epochs = 100
for i in range(epochs):
    loss_slope = -2*np.sum(y - m*X.ravel() -b)
    b = b - (lr*loss_slope)
    
    y_pred = m*X+b
    plt.plot(X,y_pred)

plt.scatter(X,y)
```
